# 2. Создание источника фотонов

Источник излучает изотропно. Световыход линейно зависит от энергии: `Nγ = Y * E`, где `Y = 10000` фотонов/МэВ. Для изотропии равномерно разыгрываются азимут `φ` и `cos(θ)`.

```python
import numpy as np

class Source:
    def __init__(self, Y=10000):
        self.Y = Y
        self.rng = np.random.default_rng()

    def generate_directions(self, E):
        if E < 0:
            raise ValueError("Энергия должна быть неотрицательной")

        photon_count = round(self.Y * E)
        cos_theta = self.rng.uniform(-1.0, 1.0, photon_count)
        phi = self.rng.uniform(0.0, 2 * np.pi, photon_count)
        sin_theta = np.sqrt(1.0 - cos_theta**2)
        return np.column_stack((
            sin_theta * np.cos(phi),
            sin_theta * np.sin(phi),
            cos_theta
        ))
```

```python
source = Source()
source.rng = np.random.default_rng(42)
directions = source.generate_directions(1.0)

print(f"Число фотонов: {len(directions)}")
print(f"Средняя длина вектора: {np.linalg.norm(directions, axis=1).mean():.6f}")
```
